🔧 OPERASI FUNDAMENTAL PANDAS

```python
import pandas as pd
import numpy as np
```

1. DATAFRAME CREATION

```python
print("=== DATAFRAME CREATION ===")
# From dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'Salary': [50000, 60000, 70000, 55000],
    'Department': ['IT', 'HR', 'IT', 'Finance']
}
df = pd.DataFrame(data)
print("DataFrame:\n", df)
```

2. ACCESS & INDEXING

```python
print("\n=== ACCESS & INDEXING ===")
print("Column 'Name':", df['Name'].tolist())
print("First row:\n", df.iloc[0])                    # By integer position
print("Rows 1-2:\n", df.iloc[1:3])
print("Filter Age > 28:\n", df[df['Age'] > 28])
```

3. ADD/REMOVE COLUMNS

```python
print("\n=== ADD/REMOVE COLUMNS ===")
df['Experience'] = [2, 5, 8, 3]                     # Add new column
df['Senior'] = df['Age'] > 30                       # Conditional column
df = df.drop('Experience', axis=1)                  # Remove column
print("After modifications:\n", df)
```

4. HANDLING MISSING VALUES

```python
print("\n=== MISSING VALUES ===")
df_with_na = df.copy()
df_with_na.loc[1, 'Salary'] = np.nan
df_with_na.loc[2, 'Age'] = np.nan

print("With missing values:\n", df_with_na)
print("Is null:\n", df_with_na.isnull())

df_with_na['Salary'] = df_with_na['Salary'].fillna(df_with_na['Salary'].mean())
df_with_na['Age'] = df_with_na['Age'].fillna(df_with_na['Age'].mean())

print("Filled DataFrame:\n", df_with_na)
```

5. GROUPING & AGGREGATION

```python
print("\n=== GROUPING & AGGREGATION ===")
grouped = df.groupby('Department')
print("Grouped means:\n", grouped['Salary'].mean())
print("Grouped description:\n", grouped['Salary'].describe())
```

6. SORTING

```python
print("\n=== SORTING ===")
df_sorted = df.sort_values('Salary', ascending=False)
print("Sorted by Salary:\n", df_sorted)
```

7. MERGING & JOINING

```python
print("\n=== MERGING ===")
df2 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Eve'],
    'Bonus': [5000, 3000, 4000]
})
merged = pd.merge(df, df2, on='Name', how='left')
print("Merged DataFrame:\n", merged)
```

8. PIVOT TABLES

```python
print("\n=== PIVOT TABLES ===")
pivot = df.pivot_table(values='Salary', index='Department', aggfunc=['mean', 'count'])
print("Pivot table:\n", pivot)
```

9. DATETIME OPERATIONS

```python
print("\n=== DATETIME OPERATIONS ===")
dates = pd.date_range('20230101', periods=4)
df_dates = df.copy()
df_dates['Join_Date'] = dates
df_dates['Year'] = df_dates['Join_Date'].dt.year
print("With dates:\n", df_dates)
```

10. STRING OPERATIONS

```python
print("\n=== STRING OPERATIONS ===")
df['Name_Upper'] = df['Name'].str.upper()
df['Name_Length'] = df['Name'].str.len()
print("String operations:\n", df[['Name', 'Name_Upper', 'Name_Length']])
```

EXPLORATORY DATA ANALYSIS

```python

import numpy as np
import pandas as pd

# ==========================================
# MEMBUAT SAMPLE RETAIL DATASET
# ==========================================

np.random.seed(42)

dates = pd.date_range(
    '2023-01-01',
    '2023-03-31',
    freq='D'
)

retail_data = pd.DataFrame({
    'date': np.random.choice(dates, 1000),
    'product_id': np.random.randint(1, 11, 1000),
    'quantity': np.random.randint(1, 10, 1000),
    'price': np.random.uniform(10, 100, 1000),
    'customer_id': np.random.randint(1, 101, 1000)
})

retail_data['revenue'] = (
    retail_data['quantity'] * retail_data['price']
)


# ==========================================
# TODO 1: TOTAL REVENUE PER PRODUCT
# ==========================================

revenue_per_product = (
    retail_data.groupby('product_id')['revenue']
    .sum()
    .sort_values(ascending=False)
)

print("1. Total Revenue per Product:")
print(revenue_per_product)


# ==========================================
# TODO 2: TOP 5 CUSTOMERS BERDASARKAN SPENDING
# ==========================================

top_customers = (
    retail_data.groupby('customer_id')['revenue']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("\n2. Top 5 Customers:")
print(top_customers)


# ==========================================
# TODO 3: REVENUE PER HARI
# ==========================================

daily_revenue = (
    retail_data.groupby('date')['revenue']
    .sum()
    .sort_index()
)

print("\n3. Revenue per Hari:")
print(daily_revenue)


# ==========================================
# TODO 4: DETECT QUANTITY ANOMALIES
# ==========================================

def detect_quantity_anomalies(df, threshold=2):

    # Menghitung rata-rata quantity
    mean = df['quantity'].mean()

    # Menghitung standar deviasi
    std = df['quantity'].std()

    # Menghitung Z-score
    z_score = (df['quantity'] - mean) / std

    # Mengambil data yang memiliki Z-score
    # lebih besar dari threshold
    anomalies = df[abs(z_score) > threshold].copy()

    # Menambahkan kolom Z-score
    anomalies['z_score'] = z_score[abs(z_score) > threshold]

    return anomalies


anomalies = detect_quantity_anomalies(retail_data)


# ==========================================
# MENAMPILKAN HASIL ANOMALIES
# ==========================================

print("\n4. Quantity Anomalies:")
print(anomalies)


# ==========================================
# VALIDASI
# ==========================================

assert len(revenue_per_product) <= 10, \
    "Should have max 10 products"

assert len(top_customers) == 5, \
    "Should have top 5 customers"

print("\n✅ Pandas operations completed")
```
